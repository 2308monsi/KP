```python
import numpy as np
import random

# Example bigram model: all real values = P(next word | prefix)
bigram_probs = {
    "<s>": {"I": 0.5, "The": 0.3, "She": 0.2},
    "I": {"like": 0.6, "am": 0.4},
    "The": {"cat": 0.5, "dog": 0.5},
    "She": {"runs": 1.0},
    "like": {"coffee": 0.7, "tea": 0.3},
    "am": {"tired": 0.5, "happy": 0.5},
    "cat": {"sleeps": 1.0},
    "dog": {"barks": 1.0},
    "runs": {"fast": 1.0},
    "coffee": {"</s>": 1.0},
    "tea": {"</s>": 1.0},
    "tired": {"</s>": 1.0},
    "happy": {"</s>": 1.0},
    "sleeps": {"</s>": 1.0},
    "barks": {"</s>": 1.0},
    "fast": {"</s>": 1.0},
}

# Randomly sample the next word based on conditional probabilities.
def sample_next_word(prefix):
    next_words = list(bigram_probs[prefix].keys())
    probs = list(bigram_probs[prefix].values())
    next_word = random.choices(next_words, weights=probs, k=1)[0]	# k is number of next-word-candidates to be generated
    #print(f"From '{prefix}' sampled next word: '{next_word}'")
    return next_word

# Generate a sentence starting from <s> and ending at </s>.
def generate_sentence():
    sentence = []
    word = "<s>"
    while True:
        next_word = sample_next_word(word)
        if next_word == "</s>":
            break
        sentence.append(next_word)
        word = next_word
    return " ".join(sentence)		# Convert list of generated words to a single string using space as separator

# Generate several random sentences
for _ in range(5):
    print(generate_sentence())
    
```
